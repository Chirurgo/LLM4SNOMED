```python
def topk_accuracy(gold_data, pred_data, top_k_values=[1, 5, 10]):
    """
    Evaluate top-k concept ranking accuracy from predicted candidates.
    Match is based on span (start, end) only.

    Args:
        gold_data: list of dicts with 'entity' and 'Target code'
        pred_data: list of list of predicted entities with 'tops'
        top_k_values: which k values to report (default: [1, 5, 10])

    Returns:
        dict with top-k accuracy scores
    """
    assert len(gold_data) == len(pred_data), "Mismatch in number of records."

    # Total number of gold entities (all, regardless of prediction)
    total_gold = sum(len(item.get("entity", [])) for item in gold_data)

    # Total number of predicted entities (all, regardless of gold)
    total_predicted = sum(len(ents or []) for ents in pred_data)

    topk_hits = {k: 0 for k in top_k_values}
    total = 0  # number of *evaluated* concepts (where pred span matches a gold span)
    empty_list_count = 0 

    for gold_item, pred_entities in zip(gold_data, pred_data):
        gold_ents = {
            (e["start_index"], e["end_index"]): str(e["Target code"])
            for e in gold_item.get("entity", [])
        }
        pred_entities = pred_entities or []
        for pred in pred_entities:
            span = (pred["start_index"], pred["end_index"])
            if span not in gold_ents:
                # predicted span with no gold counterpart → not evaluated
                continue  

            gold_cid = gold_ents[span]
            tops = pred.get("tops")

            if not tops:
                empty_list_count += 1
                top_candidates = ["Unknown"]
            else:
                top_candidates = [
                    str(t.get("concept_id")) if t and t.get("concept_id") is not None else "Unknown"
                    for t in tops
                ]

            total += 1
            for k in top_k_values:
                if gold_cid in top_candidates[:k]:
                    topk_hits[k] += 1

    empty_list_percentage = (empty_list_count / total) * 100 if total else 0.0
    unmatched_predicted = total_predicted - total  # predicted spans that never matched any gold span

    results = {
        f"top{k}": topk_hits[k] / total if total else 0.0
        for k in top_k_values
    }

    print(f"Total gold concepts: {total_gold}")
    print(f"Total predicted concepts: {total_predicted}")
    print(f"Evaluated concepts (pred span matched gold span): {total}")
    print(f"Predicted spans with no gold match: {unmatched_predicted}")
    print(f"No. of no mappable concept (among evaluated): {empty_list_count} ({empty_list_percentage:.2f}%)")

    return results
```

```python
import json
with open(path_testdata, "r", encoding="utf-8") as f:
    gold_data = json.load(f)
predict_path = pred_repo + "\\medgemma.json"
with open(predict_path, "r", encoding="utf-8") as f:
    pred_data = json.load(f)

scores = topk_accuracy(gold_data, pred_data)
print("<<< Concept Precision, Recall, F1 Evaluation>>>")
```
